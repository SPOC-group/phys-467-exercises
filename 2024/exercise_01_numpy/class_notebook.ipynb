Linear Algebra

```python
A = np.random.randn(3,3)
b = np.random.randn(3)
```

```python
A @ b, np.matmul(A, b)
```

```python
np.dot(b, b)
```

```python
M.T
```

```python
eigenvalues, eigenvectors = np.linalg.eig(M)
eigenvalues, eigenvectors
```

```python
eigenvectors[:,0]
```

```python
M @ eigenvectors[:,0] == eigenvalues[0] * eigenvectors[:,0]
```

```python
np.isclose(M @ eigenvectors[:,0], eigenvalues[0] * eigenvectors[:,0])
```

Einsum

```python
v = np.array([7, 5, 9])
v
```

```python
np.einsum('i,i', v, v) # v @ v
```

```python
M
```

```python
np.einsum('ij,jk', M, M) # np.matmul(M, M)
```

```python
np.einsum('ij,jh->ih', M, M)
```

```python
np.einsum('ij,j', M, v) # np.matmul(M, v)
```

```python
np.einsum('ii', M) # np.trace(M)
```

```python
np.einsum('ji', M) # M.T
```
