# Exercise 3: QHO

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_min = -10
x_max = 10
dx = 0.01
x = np.arange(x_min, x_max+dx, dx)
N = len(x)
x
```

```python
K = np.diag(-2 * np.ones(len(x)), k=0) + np.diag(np.ones(len(x)-1), k=1) + np.diag(np.ones(len(x)-1), k=-1)

K = -.5 * K / dx**2
```

```python
V = .5 * x**2

plt.plot(x, V, 'k')
plt.xlabel("x")
plt.ylabel("V(x)")

V = np.mat(np.diag(V))
```

```python
H = K + V
```

```python
%%timeit
eval, evec = np.linalg.eig(H)
```

```python
%%timeit
eval, evec = np.linalg.eigh(H)
```

```python
eval, evec = np.linalg.eigh(H)
# sorted_indices = np.argsort(eval)
# eval, evec = eval[sorted_indices], evec[:,sorted_indices]
```

```python
normalization = np.sqrt(np.multiply(np.conjugate(evec), evec).sum(axis=0) * dx)
psi = evec / normalization

for i in range(10):
    print(f"\nn = {i}, E_num = {eval[i]:.4f}, E_th = {i + .5}")
    plt.plot(x, psi[:, i], 'k')
    plt.xlabel("x")
    plt.ylabel(rf"$\psi_{i}(x)$")
    plt.show()
```

# Exercise 4: Random Matrices

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 1000
A = np.random.randn(N, N)
M = (A + A.T) / (2 * N)**.5
```

```python
eval, evec = np.linalg.eigh(M)
```

```python
plt.hist(eval, bins=100, density=True, alpha=.8)
x = np.linspace(-2, 2, 100)
y = (4 - x**2)**.5 / (2*np.pi)
plt.plot(x, y, 'k')
plt.xlabel(r"$\lambda$")
plt.ylabel(r"$\rho(\lambda)$")
```
